```python
#7.Write a Python script to draw a boxplot for the following numeric list and identify its outliers. Explain the result:data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]

import numpy as np
import matplotlib.pyplot as plt

data = [12, 14, 14, 15, 18, 19, 19, 21,
        22, 22, 23, 23, 24, 26, 29, 35]

# Calculate quartiles and IQR
Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1

# Calculate lower and upper fences
lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR

# Identify outliers
outliers = [x for x in data
            if x < lower_fence or x > upper_fence]

# Print results
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower fence:", lower_fence)
print("Upper fence:", upper_fence)
print("Outliers:", outliers)

# Draw boxplot
plt.boxplot(data)
plt.title("Boxplot and Outlier Detection")
plt.ylabel("Values")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```

You are working as a data analyst in an e-commerce company. The marketing team wants to know if there is a relationship between advertising spend and daily sales.
>>Explain how you would use covariance and correlation to explore this relationship.

>>Ans...Covariance: If covariance between advertising spend and daily sales is positive, higher advertising spend tends to be associated with higher sales.

>>Correlation: Tells us how strongly advertising spend and daily sales move together.
>-  +1: Perfect positive linear relationship.
>- 0: No linear relationship.
>- -1: Perfect negative linear relationship.


```python
#8.Write Python code to compute the correlation between the two lists: advertising_spend = [200, 250, 300, 400, 500] daily_sales = [2200, 2450, 2750, 3200, 4000]

import numpy as np
advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]
correlation_matrix = np.corrcoef(advertising_spend, daily_sales)
correlation = correlation_matrix[0, 1]
print("Correlation:", correlation)
```
